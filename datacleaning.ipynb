```python
import pandas as pd   
import numpy as np
```

```python
dataset=pd.read_csv('data.csv',lineterminator='\n')
dataset.head(4)
```

```python
dataset.info()
```

```python
dataset['Release_Date']=pd.to_datetime(dataset['Release_Date'])
print(dataset['Release_Date'].dtypes)
```

```python
dataset['Release_Date']=pd.DatetimeIndex(dataset['Release_Date']).year
```

```python
dataset.duplicated().sum()
```

```python
col=['Overview','Original_Language','Poster_Url']
dataset.drop(col,axis=1,inplace=True)
dataset.columns
```

```python
dataset.head(5)
```

```python
def categorize_col(datasett,col,lab):
    edes=[
        datasett[col].describe()['min'],
        datasett[col].describe()['25%'],
        dataset[col].describe()['50%'],
        datasett[col].describe()['75%'],
        datasett[col].describe()['max'],
        ]
    
    datasett[col]=pd.cut(datasett[col],edes,labels=lab,duplicates='drop')
    return datasett
```

```python
labb=['not_good','below_avg','avg','good']
categorize_col(dataset,'Vote_Average',labb)
```

```python
dataset['Vote_Average'].unique()
```

```python
dataset['Vote_Average'].value_counts()
```

```python
dataset.isnull().sum()
dataset.dropna(inplace=True)
```

```python
dataset['Genre']=dataset['Genre'].str.split(', ')
dataset=dataset.explode('Genre').reset_index(drop=True)
```

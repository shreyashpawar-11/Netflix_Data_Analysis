```python
import pandas as pd   
import matplotlib.pyplot as plt   
import seaborn as sns             
import numpy as np          
```

```python
df=pd.read_csv('data_clean.csv')
df.head()
```

What is the most frequent genre of movies released on Netflix?

```python
#plt.figure(figsize=(27,6))
sns.histplot(y=df['Genre'])
plt.show()
```

```python
df['Genre'].value_counts().head()
```

What genre has a high average rating?

```python
li=df['Vote_Average'].value_counts().index
li
```

```python
sns.catplot(x="Vote_Average",  data=df, kind="count",order=df['Vote_Average'].value_counts().index)
plt.show()
```

What movie got the highest popularity? what's its genre?

```python
print(df['Popularity'].max())
var=df[df['Popularity']==df['Popularity'].max()]
var[['Genre','Popularity','Title']]
```

What year has the lowest popularity? what's its genre?

```python
print(df['Popularity'].max())
var=df[df['Popularity']==df['Popularity'].min()]
var[['Release_Date','Genre','Popularity','Title']]
```

Which year has the most filmmed movies?

```python
plt.figure(figsize=(14,6))
sns.histplot(x='Release_Date',data=df)
plt.plot()
```
